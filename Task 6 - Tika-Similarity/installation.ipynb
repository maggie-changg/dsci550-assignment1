# Task 6d: Convert TSV dataset into JSON using ETLLib’s tsv2json tool
Once ETLLib is installed, we want to use the tsvtojson.py tool to convert the 'haunted_place.tsv' dataset to 'haunted_place.json' by using the command: </br>
**PYTHONPATH=etllib python etllib/etl/tsvtojson.py \ -t ../data/haunted_places.tsv \ -j "../Task 6 - Tika-Similarity/data/json/haunted_places.json" \ -c ../data/column_headers.txt \ -o hauntedPlace \ -s 0.50**

The JSON file can be recreated using the following steps: </br>

**Step 1:** Generate 'column_headers.txt' by running the script below to automatically extract headers from the 'haunted_place.tsv' file to save them into the data/column_headers.txt </br>
This code below reads the first line of 'haunted_places.tsv' and writes each column name into its own line in 'column_headers.txt'. 

**Step 2:** Run the TSV to JSON conversion using the 'tsvtojson.py' script inside etllib to convert 'haunted_places.tsv' into 'haunted_places.json' using the command above. 





```python
import os

# Ensure Data folder exists (for the general data folder, move up one level)
os.makedirs('../data', exist_ok=True)

# Read TSV headers from the general data folder
with open('../data/haunted_places.tsv', 'r', encoding='utf-8') as f:
    header_line = f.readline().strip()

# Split headers by tab character (TSV format)
headers = header_line.split('\t')

# Write each header to a new line in the general data/column_headers.txt file
with open('../data/column_headers.txt', 'w', encoding='utf-8') as f:
    for header in headers:
        f.write(header + '\n')

# Confirmation message
print("column_headers.txt has been created in the general 'data' folder.")
```

# Splitting Aggregate JSON into Individual JSON Files
After converting the `haunted_places.tsv` file into a single **aggregate JSON file (`haunted_places.json`)**, we now need to **split** this file so that each **row** becomes an **individual JSON file**.

## **How This Works**
- The script reads the **aggregate JSON file** (`haunted_places.json`).
- It creates a new directory called **`individual_jsons/`** inside the `Task 6 - Tika-Similarity/data/json/` folder.
- It **loops through each row (feature) in the JSON** and saves it as a separate JSON file.
- Each JSON file is named **`haunted_place_1.json`**, `haunted_place_2.json`, etc.
- The script ensures that the **output directory exists** before writing files.

```python
import json
import os
```

```python
# Set Paths
input_json_path = "../Task 6 - Tika-Similarity/data/json/haunted_places.json"

output_directory = "../Task 6 - Tika-Similarity/data/json/individual_jsons/"
```

```python
# Load the aggregated JSON file
with open(input_json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Extract the list of haunted places
haunted_places = data.get("hauntedPlace", [])

# Ensure output directory exists
output_dir = output_directory
os.makedirs(output_dir, exist_ok=True)

# Loop through each entry and write to an individual JSON file
for i, entry in enumerate(haunted_places):
    filename = os.path.join(output_dir, f"haunted_place_{i+1}.json")
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(entry, f, indent=4)

print(f"Successfully created {len(haunted_places)} individual JSON files in '{output_dir}' folder.")
```
