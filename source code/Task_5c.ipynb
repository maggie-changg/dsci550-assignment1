```python
import pandas as pd

# Load datasets
merged_haunted_df = pd.read_csv("../data/merged_haunted_places_2.csv")
substance_df = pd.read_csv("../data/substance.csv")
```

```python
# Standardize "State" column format in both datasets
merged_haunted_df["State"] = merged_haunted_df["State"].str.strip().str.title()
substance_df["State"] = substance_df["State"].str.strip().str.title()
```

```python
# Merge substance data with haunted data
merged_df = merged_haunted_df.merge(substance_df, on="State", how="left")
```

```python
# Rename substance columns for clarity
merged_df.rename(columns={
    "Top_6_Primary_Substances": "Top_6_Primary_Substances (Per 100,000)",
    "Alcohol": "Alcohol (Per 100,000)",
    "Heroin": "Heroin (Per 100,000)",
    "Methamphetamine": "Methamphetamine (Per 100,000)",
    "Other_Opiates_Synthetics": "Other Opiates/Synthetics (Per 100,000)",
    "Marijuana_Hashish": "Marijuana/Hashish (Per 100,000)",
    "Cocaine": "Cocaine (Per 100,000)"
}, inplace=True)
```

```python
# Save the Final Merged Dataset
merged_df.to_csv("merged_haunted_places_all_3.csv", index=False)

print(merged_df.head())
print(merged_df.shape)
```
