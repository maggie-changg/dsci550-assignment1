# Task 6d: Convert TSV dataset into JSON using ETLLib’s tsv2json tool
Once ETLLib is installed, we want to use the tsvtojson.py tool to convert the 'haunted_place.tsv' dataset to 'haunted_place.json' by using the command: </br>
**PYTHONPATH=etllib python etllib/etl/tsvtojson.py -t haunted_places.tsv -j data/haunted_places.json -c data/column_headers.txt -o hauntedPlace -s 0.5**

The JSON file can be recreated using the following steps: </br>

**Step 1:** Generate 'column_headers.txt' by running the script below to automatically extract headers from the 'haunted_place.tsv' file to save them into the data/column_headers.txt </br>
This code below reads the first line of 'haunted_places.tsv' and writes each column name into its own line in 'column_headers.txt'. 

**Step 2:** Run the TSV to JSON conversion using the 'tsvtojson.py' script inside etllib to convert 'haunted_places.tsv' into 'haunted_places.json' using the command above. 





```python
import os

# Ensure Data folder exists
os.makedirs('data', exist_ok=True)

# Read TSV headers
with open('data/haunted_places.tsv', 'r', encoding='utf-8') as f:
    header_line = f.readline().strip()

# Split headers by tab character (TSV format)
headers = header_line.split('\t')

# Write each header to a new line in Data/column_headers.txt
with open('data/column_headers.txt', 'w', encoding='utf-8') as f:
    for header in headers:
        f.write(header + '\n')
        
# Confirmation that the 'column_headers.txt' file is successfully created. 
print("column_headers.txt has been created in the 'Data' folder.")
```

# Preparing Directory to use for Tika Similarity

```python
import os
import json

# Ensure output directory exists
output_dir = 'data/haunted_places_jsons'
os.makedirs(output_dir, exist_ok=True)

# Load the JSON dataset
with open('data/haunted_places.json', 'r') as f:
    data = json.load(f)

# Load the headers from column_headers.txt
with open('data/column_headers.txt', 'r', encoding='utf-8') as f:
    headers = [line.strip() for line in f]

# Split the JSON data into individual files
for i, entry in enumerate(data['hauntedPlace']):
    row_data = {header: entry.get(header, None) for header in headers}
    output_file = os.path.join(output_dir, f'haunted_place_{i+1}.json')
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(row_data, f, indent=4)

print(f"Split {len(data['hauntedPlace'])} entries into individual JSON files in the '{output_dir}' directory.")
```

```python
import os
import json

# Ensure output directory exists
output_dir = 'data/haunted_places_jsons'
os.makedirs(output_dir, exist_ok=True)

# Load the JSON dataset
with open('data/haunted_places.json', 'r') as f:
    data = json.load(f)

# Split the JSON data into individual files
for i, entry in enumerate(data):
    output_file = os.path.join(output_dir, f'haunted_place_{i+1}.json')
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(entry, f, indent=4)

print(f"Split {len(data)} entries into individual JSON files in the '{output_dir}' directory.")
```

Subsetting 700 files to use for Tika Similarity

```python
import shutil

# Ensure new output directory exists
new_output_dir = 'data/haunted_places_jsons_subset'
os.makedirs(new_output_dir, exist_ok=True)

# Get list of JSON files in the original directory
json_files = [f for f in os.listdir(output_dir) if f.endswith('.json')]

# Copy the first 700 JSON files to the new directory
for json_file in json_files[:700]:
    shutil.copy(os.path.join(output_dir, json_file), os.path.join(new_output_dir, json_file))

print(f"Copied 700 JSON files to the '{new_output_dir}' directory.")
```
