```python
import re

# Extract state abbreviation using regex
daylight_df['state_abbrev'] = daylight_df['state_abbrev'].str.extract(r'([A-Z]{2})', expand=False)
print(daylight_df['state_abbrev'].head())
```

```python
print(repr(merged_haunted_df.columns.tolist()))
print(repr(daylight_df.columns.tolist()))
```

```python

final_merged_df = merged_haunted_df.merge(daylight_df, on="state_abbrev", how="left")
```

```python
# Save the final merged dataset
print("Final merged dataset shape:", final_merged_df.shape)
print(final_merged_df.head())
final_merged_df.to_csv("../data/merged_haunted_places_3.csv", index=False, header=True)
```
